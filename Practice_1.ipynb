```python
import pandas as pd
```

```python
df = pd.read_csv("practice_daraz_dataset.csv")
```

```python
df["price"] = pd.to_numeric(df["price"], errors = "coerce") #converts price into numeric and set errors to NaN 
df["price"] = df["price"].fillna(df["price"].mean())
df
```

```python
df["age"] = pd.to_numeric(df["age"], errors = "coerce")
df["age"] = df["age"].fillna(df["age"].median()).astype(int)
df
```

```python
df["discount_percent"]
```

```python
df["discount_percent"] = df["discount_percent"].str.replace("%","") #replaces % with the given
df
```

```python
df.duplicated() # checks the duplicated i.e true only if everything is same
```

```python
df.duplicated("product_id") #checks the duplicate in the specific column
```

```python
df.drop_duplicates(inplace=True) #removes the duplicates while inplace = True ensures the data to be updated
df
```

```python
df.drop_duplicates(subset="product_id", inplace = True) #removes the duplicated in the specific column ; keep = first or last can be given too
df
```

```python
df.loc[df["price"]<0,"price"]=0 #if price has negative value then set to 0
df
```

```python
df["city"]=df["city"].str.strip() #removes the whitespace from both sides ; lstrip() and rstrip() are for left and right respectively
df["city"]
```

```python
df["city"].str.lower() #changes into lower case similarly upper() for upper case
```
