---

## 🔍 Introduction

An outlier is a data point that differs significantly from other observations in a dataset. It represents an unusually high or low value compared to the general pattern of the data.

In Artificial Intelligence and Machine Learning, outliers are important because they can influence statistical calculations, affect model training, and reduce prediction accuracy.

Outliers may occur due to human errors, incorrect measurements, data entry mistakes, system failures, or naturally unusual observations.

## 🧠 Understanding Outliers

Outliers are observations that fall far away from the typical distribution of data. They may appear in numerical datasets as extremely large or small values.

For example, if the ages in a dataset are 18, 20, 21, 22, 23, and 150, the value 150 may be considered an outlier because it differs considerably from the other observations.

## ⚠️ Types of Outliers

### Univariate Outliers

Univariate outliers are unusual values identified by examining a single feature or variable.

### Multivariate Outliers

Multivariate outliers are unusual observations identified by examining relationships between two or more variables.

## 📈 Causes of Outliers

Outliers can be caused by incorrect data entry, measurement errors, faulty sensors, unusual human behavior, exceptional events, or natural variations in data.

## 🛠️ Detecting Outliers

Outliers can be detected using statistical techniques such as the Interquartile Range (IQR), Z-score, and visualization methods such as box plots and scatter plots.

## 📐 Interquartile Range (IQR) Method

The Interquartile Range method is a statistical technique used to detect outliers by examining the spread of the middle 50% of the dataset.

**IQR = Q3 − Q1**

**Lower Bound = Q1 − 1.5 × IQR**

**Upper Bound = Q3 + 1.5 × IQR**

Values below the lower boundary or above the upper boundary are generally considered potential outliers.

## 🤖 Impact of Outliers on Machine Learning

Outliers can influence machine learning algorithms, especially those that depend on distances, averages, or statistical distributions.

Linear Regression and K-Nearest Neighbors can be sensitive to extreme values, while Decision Trees and Random Forests are generally less sensitive to numerical outliers.

## 🧹 Handling Outliers

Outliers can be handled by removing incorrect observations, applying transformations, using capping techniques, or retaining meaningful extreme values.

The appropriate method depends on the dataset and the problem being solved.

```python
import pandas as pd
df = pd.read_csv("practice_daraz_dataset.csv")
df
```

```python
df["price"] = pd.to_numeric(df["price"], errors = "coerce")
df
```

```python
#Finding Quartiles for IQR
Q1 = df["price"].quantile(0.25)
Q2 = df["price"].quantile(0.50)
Q3 = df["price"].quantile(0.75)

#Finding IQR
IQR = Q3 - Q1

#Finding lower and upper bound
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)
```

```python
#Checking the data for the outliers
Outliers = df[(df["price"]<lower_bound) | (df["price"]>upper_bound)]
Outliers
```

```python
len(Outliers) #gives the number of outliers
```

```python
df.loc[(df["price"]<lower_bound) | (df["price"]>upper_bound), "price"] = Q2 #setting the value to Median ; df.loc[rows, column]
df
```

```python
df["price"] = df["price"].clip(lower_bound, upper_bound) #sets the lower_bound value if less than lower_bound else sets upper_bound for the values exceeding upperbound
df["price"]
```
